**Step-8:Drop dataframes**

```python
import pandas as pd
names = ["Ramesh","Suresh","Mahesh"]
age = [20,22,24]
city = ["hyd","vishka","ibp"]
df = pd.DataFrame(zip(names,age,city),columns= ["Names","Age","City"])
df
df.drop(age)
df.drop(age,axis=1)
```

**Step-9:Rename the columns**

```python
import pandas as pd
names = ["Ramesh","Suresh","Mahesh"]
age = [20,22,24]
city = ["hyd","vishka","ibp"]
df = pd.DataFrame(zip(names,age,city),columns= ["Names","Age","City"])
df

```

```python
dict1 = {"City":"city"}
df.rename(columns=dict1,inplace=True)
df
```

```python
dict2 = {2:'B'}
df.rename(dict2,inplace=True)
df
```
